# Lab 03 - Discrete Filters and Edge Detection\n
\n
This notebook solves the 3 requested tasks using functions from `tools.py`:\n
1. Grayscale convolution experiments (padding/stride/dilation + comparison).\n
2. Automation with an averaging-filter builder and analysis function.\n
3. Repeat the experiment for a color image.

```python
import cv2\n
import numpy as np\n
import matplotlib.pyplot as plt\n
from pathlib import Path\n
\n
import tools
```

```python
def find_flower_image():\n
    candidates = [\n
        Path("flower.jpg"),\n
        Path("flower.png"),\n
        Path("../lab-1/flower.jpg"),\n
        Path("../lab-1/flower.png"),\n
    ]\n
    for p in candidates:\n
        if p.exists():\n
            return p\n
    raise FileNotFoundError("No flower image found in expected paths.")\n
\n
\n
img_path = find_flower_image()\n
img_gray = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)\n
img_bgr = cv2.imread(str(img_path), cv2.IMREAD_COLOR)\n
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)\n
\n
print("Image path:", img_path)\n
print("Gray shape:", img_gray.shape)\n
print("Color shape:", img_rgb.shape)
```

```python
def averaging_mask(size):\n
    """Build a square averaging kernel of shape (size, size)."""\n
    if size < 1 or size % 2 == 0:\n
        raise ValueError("size must be a positive odd integer")\n
    kernel = np.ones((size, size), dtype=np.float32)\n
    return kernel / kernel.sum()\n
\n
\n
def apply_filter_single_channel(img, kernel, stride=(1, 1), dilation=(1, 1), padding=(0, 0)):\n
    return tools.Conv2D(\n
        img.astype(np.float32),\n
        np.asarray(kernel, dtype=np.float32),\n
        stride=stride,\n
        dilation=dilation,\n
        padding=padding,\n
    )\n
\n
\n
def apply_filter_color(img_rgb, kernel, stride=(1, 1), dilation=(1, 1), padding=(0, 0)):\n
    channels = [\n
        apply_filter_single_channel(img_rgb[:, :, c], kernel, stride, dilation, padding)\n
        for c in range(3)\n
    ]\n
    out = np.dstack(channels)\n
    return np.clip(out, 0, 255).astype(np.uint8)
```

## 1) Grayscale Experiment: Padding, Stride, Dilation, and Mixed Variants

```python
kernel_sizes = [3, 7, 11]\n
stride_values = [2, 5, 11]\n
dilation_values = [1, 3, 5]\n
\n
results_gray = []\n
\n
# Original image\n
results_gray.append(("Original", img_gray))\n
\n
# Averaging with padding to keep original dimensions\n
for k in kernel_sizes:\n
    ker = averaging_mask(k)\n
    p = (k // 2, k // 2)\n
    out = apply_filter_single_channel(img_gray, ker, padding=p)\n
    results_gray.append((f"Avg k={k}, pad={p} -> {out.shape}", out))\n
\n
# Stride variants without padding (k=3)\n
k_base = 3\n
ker_base = averaging_mask(k_base)\n
for s in stride_values:\n
    out = apply_filter_single_channel(img_gray, ker_base, stride=(s, s), padding=(0, 0))\n
    results_gray.append((f"Stride=({s},{s}), no pad -> {out.shape}", out))\n
\n
# Dilation variants without padding (k=3)\n
for d in dilation_values:\n
    out = apply_filter_single_channel(img_gray, ker_base, dilation=(d, d), padding=(0, 0))\n
    results_gray.append((f"Dilation=({d},{d}), no pad -> {out.shape}", out))\n
\n
# Combine some variants\n
combo1 = apply_filter_single_channel(\n
    img_gray, ker_base, stride=(2, 2), dilation=(3, 3), padding=(0, 0)\n
)\n
combo2 = apply_filter_single_channel(\n
    img_gray, ker_base, stride=(5, 5), dilation=(1, 1), padding=(1, 1)\n
)\n
results_gray.append((f"Combo s=(2,2), d=(3,3), no pad -> {combo1.shape}", combo1))\n
results_gray.append((f"Combo s=(5,5), d=(1,1), p=(1,1) -> {combo2.shape}", combo2))\n
\n
n = len(results_gray)\n
cols = 4\n
rows = int(np.ceil(n / cols))\n
fig, axes = plt.subplots(rows, cols, figsize=(18, 4 * rows))\n
axes = np.array(axes).reshape(-1)\n
\n
for i, (title, im) in enumerate(results_gray):\n
    axes[i].imshow(im, cmap="gray")\n
    axes[i].set_title(title, fontsize=9)\n
    axes[i].axis("off")\n
\n
for j in range(i + 1, len(axes)):\n
    axes[j].axis("off")\n
\n
plt.tight_layout()\n
plt.show()
```

## 2) Automating the Process with a Generic Analysis Function

```python
def filter_analysis(img, filter_builder, sizes=(3, 7, 11), stride_vals=(2, 5, 11), dilation_vals=(1, 3, 5)):\n
    """Return a comparison figure for one-channel image and multiple convolution variants."""\n
    outputs = [("Original", img)]\n
\n
    for k in sizes:\n
        ker = filter_builder(k)\n
        p = (k // 2, k // 2)\n
        out = apply_filter_single_channel(img, ker, padding=p)\n
        outputs.append((f"Avg k={k}, pad={p} -> {out.shape}", out))\n
\n
    ker3 = filter_builder(3)\n
    for s in stride_vals:\n
        out = apply_filter_single_channel(img, ker3, stride=(s, s), padding=(0, 0))\n
        outputs.append((f"Stride=({s},{s}), no pad -> {out.shape}", out))\n
\n
    for d in dilation_vals:\n
        out = apply_filter_single_channel(img, ker3, dilation=(d, d), padding=(0, 0))\n
        outputs.append((f"Dilation=({d},{d}), no pad -> {out.shape}", out))\n
\n
    outputs.append((\n
        f"Combo s=(2,2), d=(3,3), no pad -> {apply_filter_single_channel(img, ker3, stride=(2,2), dilation=(3,3), padding=(0,0)).shape}",\n
        apply_filter_single_channel(img, ker3, stride=(2, 2), dilation=(3, 3), padding=(0, 0)),\n
    ))\n
\n
    n = len(outputs)\n
    cols = 4\n
    rows = int(np.ceil(n / cols))\n
    fig, axes = plt.subplots(rows, cols, figsize=(18, 4 * rows))\n
    axes = np.array(axes).reshape(-1)\n
\n
    for i, (title, out) in enumerate(outputs):\n
        axes[i].imshow(out, cmap="gray")\n
        axes[i].set_title(title, fontsize=9)\n
        axes[i].axis("off")\n
\n
    for j in range(i + 1, len(axes)):\n
        axes[j].axis("off")\n
\n
    plt.tight_layout()\n
    return fig\n
\n
\n
fig = filter_analysis(img_gray, averaging_mask)\n
plt.show()
```

## 3) Repeat the Same Experiment for the Color Image

```python
def filter_analysis_color(img_rgb, filter_builder, sizes=(3, 7, 11), stride_vals=(2, 5, 11), dilation_vals=(1, 3, 5)):\n
    """Return a comparison figure for color image and multiple convolution variants."""\n
    outputs = [("Original", img_rgb)]\n
\n
    for k in sizes:\n
        ker = filter_builder(k)\n
        p = (k // 2, k // 2)\n
        out = apply_filter_color(img_rgb, ker, padding=p)\n
        outputs.append((f"Avg k={k}, pad={p} -> {out.shape}", out))\n
\n
    ker3 = filter_builder(3)\n
    for s in stride_vals:\n
        out = apply_filter_color(img_rgb, ker3, stride=(s, s), padding=(0, 0))\n
        outputs.append((f"Stride=({s},{s}), no pad -> {out.shape}", out))\n
\n
    for d in dilation_vals:\n
        out = apply_filter_color(img_rgb, ker3, dilation=(d, d), padding=(0, 0))\n
        outputs.append((f"Dilation=({d},{d}), no pad -> {out.shape}", out))\n
\n
    outputs.append((\n
        f"Combo s=(2,2), d=(3,3), no pad -> {apply_filter_color(img_rgb, ker3, stride=(2,2), dilation=(3,3), padding=(0,0)).shape}",\n
        apply_filter_color(img_rgb, ker3, stride=(2, 2), dilation=(3, 3), padding=(0, 0)),\n
    ))\n
\n
    n = len(outputs)\n
    cols = 4\n
    rows = int(np.ceil(n / cols))\n
    fig, axes = plt.subplots(rows, cols, figsize=(18, 4 * rows))\n
    axes = np.array(axes).reshape(-1)\n
\n
    for i, (title, out) in enumerate(outputs):\n
        if out.ndim == 2:\n
            axes[i].imshow(out, cmap="gray")\n
        else:\n
            axes[i].imshow(out)\n
        axes[i].set_title(title, fontsize=9)\n
        axes[i].axis("off")\n
\n
    for j in range(i + 1, len(axes)):\n
        axes[j].axis("off")\n
\n
    plt.tight_layout()\n
    return fig\n
\n
\n
fig = filter_analysis_color(img_rgb, averaging_mask)\n
plt.show()
```
