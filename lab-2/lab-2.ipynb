```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def OpenImage(path):
    I = cv2.imread(path)
    if I is None:
        raise FileNotFoundError("")
    L, C = I.shape[:2]
    return I, L, C
```

```python
def Divide(I):
    if len(I.shape) == 3:
        B, G, R = cv2.split(I)
        return B, G, R
```

```python
def HSV(I):
    return cv2.cvtColor(I, cv2.COLOR_BGR2HSV)
```

```python
def CountPix(I):
    L, C = I.shape[:2]
    N = L * C
    return N
```

```python
def FactPix(I, alpha, beta):
    new_image = cv2.convertScaleAbs(I, alpha=alpha, beta=beta)
    return new_image
```

```python
def ViewImage(I):
    if I is None:
        print(f"Error: Could not find image at {path}")
        return
    cv2.imshow("My Image Viewer", I)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
import numpy as np

def Func_a(image):
    img = image.astype(np.float32)

    log_img = np.log1p(img)
    exp_img = np.exp(img / 255)
    square_img = np.square(img)
    sqrt_img = np.sqrt(img)

    return log_img, exp_img, square_img, sqrt_img
```

```python
def Func_m(image):
    mu = np.mean(image)
    sigma = np.std(image)
    return mu, sigma
```

```python
def Normalize(image, a, b):
    norm = cv2.normalize(image, None, a, b, cv2.NORM_MINMAX)
    return norm
```

```python
def Inverse(image):
    Inv = 255 - image
    return Inv
```

```python
def CalcHist(image):
    H = cv2.calcHist([image], [0], None, [256], [0,256])
    b = np.arange(256)
    return H, b
```

```python
def Threshold(image, s):
    _, T = cv2.threshold(image, s, 255, cv2.THRESH_BINARY)
    return T
```

```python
def Func_j(path):
    img = cv2.imread(path, 0)

    plt.imshow(img, cmap="gray")
    plt.title("Original Image")
    plt.show()

    H, _ = CalcHist(img)
    plt.plot(H)
    plt.title("Original Histogram")
    plt.show()

    inv = Inverse(img)

    plt.imshow(inv, cmap="gray")
    plt.title("Inverse Image")
    plt.show()

    H2, _ = CalcHist(inv)
    plt.plot(H2)
    plt.title("Histogram After Inverse")
    plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def Func_t(path):
    img = cv2.imread(path, 0)

    plt.figure()
    plt.imshow(img, cmap="gray")
    plt.title("Original Image")
    plt.show()

    H1, b1 = CalcHist(img)
    plt.figure()
    plt.plot(b1, H1)
    plt.title("Histogram - Original")
    plt.show()

    norm = Normalize(img, 10, 50)

    plt.figure()
    plt.imshow(norm, cmap="gray")
    plt.title("Normalized Image [10,50]")
    plt.show()

    H2, b2 = CalcHist(norm)
    plt.figure()
    plt.plot(b2, H2)
    plt.title("Histogram - Normalized")
    plt.show()

    return img, norm
```

```python
img = cv2.imread("red_bird.png")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
R = img_rgb[:, :, 0]
G = img_rgb[:, :, 1]
B = img_rgb[:, :, 2]
```

```python
R
```

```python
plt.imshow(R)
```

```python
mask_rgb = (R > 150) & (G < 100) & (B < 100)
```

```python
mask_rgb
```

```python
mask_rgb = mask_rgb.astype(np.uint8) * 255
```

```python
mask_rgb
```

```python
result_rgb = cv2.bitwise_and(img_rgb, img_rgb, mask=mask_rgb)
```

```python
img_hsv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)
```

```python
lower_red1 = np.array([0, 120, 50])
upper_red1 = np.array([10, 255, 255])
lower_red2 = np.array([170, 120, 50])
upper_red2 = np.array([180, 255, 255])
```

```python
mask_hsv1 = cv2.inRange(img_hsv, lower_red1, upper_red1)
mask_hsv2 = cv2.inRange(img_hsv, lower_red2, upper_red2)
mask_hsv = cv2.bitwise_or(mask_hsv1, mask_hsv2)
result_hsv = cv2.bitwise_and(img_rgb, img_rgb, mask=mask_hsv)
```
