```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def OpenImage(path):
    I = cv2.imread(path)
    if I is None:
        raise FileNotFoundError("")
    L, C = I.shape[:2]
    return I, L, C
```

```python
def Divide(I):
    if len(I.shape) == 3:
        B, G, R = cv2.split(I)
        return B, G, R
```

```python
def HSV(I):
    return cv2.cvtColor(I, cv2.COLOR_BGR2HSV)
```

```python
def CountPix(I):
    L, C = I.shape[:2]
    N = L * C
    return N
```

```python
def FactPix(I, alpha, beta):
    new_image = cv2.convertScaleAbs(I, alpha=alpha, beta=beta)
    return new_image
```
