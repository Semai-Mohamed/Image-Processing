```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def OpenImage(path):
    I = cv2.imread(path)
    if I is None:
        raise FileNotFoundError("")
    L, C = I.shape[:2]
    return I, L, C
```

```python
def Divide(I):
    if len(I.shape) == 3:
        B, G, R = cv2.split(I)
        return B, G, R
```

```python
def HSV(I):
    return cv2.cvtColor(I, cv2.COLOR_BGR2HSV)
```

```python
def CountPix(I):
    L, C = I.shape[:2]
    N = L * C
    return N
```

```python
def FactPix(I, alpha, beta):
    new_image = cv2.convertScaleAbs(I, alpha=alpha, beta=beta)
    return new_image
```

```python
import numpy as np

def Func_a(image):
    img = image.astype(np.float32)

    log_img = np.log1p(img)
    exp_img = np.exp(img / 255)
    square_img = np.square(img)
    sqrt_img = np.sqrt(img)

    return log_img, exp_img, square_img, sqrt_img
```

```python
def Func_m(image):
    mu = np.mean(image)
    sigma = np.std(image)
    return mu, sigma
```

```python
def Normalize(image, a, b):
    norm = cv2.normalize(image, None, a, b, cv2.NORM_MINMAX)
    return norm
```

```python
def Inverse(image):
    Inv = 255 - image
    return Inv
```

```python
def CalcHist(image):
    H = cv2.calcHist([image], [0], None, [256], [0,256])
    b = np.arange(256)
    return H, b
```

```python
def Threshold(image, s):
    _, T = cv2.threshold(image, s, 255, cv2.THRESH_BINARY)
    return T
```

```python
def Func_j(path):
    img = cv2.imread(path, 0)

    plt.imshow(img, cmap="gray")
    plt.title("Original Image")
    plt.show()

    H, _ = CalcHist(img)
    plt.plot(H)
    plt.title("Original Histogram")
    plt.show()

    inv = Inverse(img)

    plt.imshow(inv, cmap="gray")
    plt.title("Inverse Image")
    plt.show()

    H2, _ = CalcHist(inv)
    plt.plot(H2)
    plt.title("Histogram After Inverse")
    plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def Func_t(path):
    img = cv2.imread(path, 0)

    # 2. Show original image
    plt.figure()
    plt.imshow(img, cmap="gray")
    plt.title("Original Image")
    plt.show()

    # 3. Histogram original
    H1, b1 = CalcHist(img)
    plt.figure()
    plt.plot(b1, H1)
    plt.title("Histogram - Original")
    plt.show()

    # 4. Normalize to [10, 50]
    norm = Normalize(img, 10, 50)

    # 5. Show normalized image
    plt.figure()
    plt.imshow(norm, cmap="gray")
    plt.title("Normalized Image [10,50]")
    plt.show()

    # 6. Histogram after normalization
    H2, b2 = CalcHist(norm)
    plt.figure()
    plt.plot(b2, H2)
    plt.title("Histogram - Normalized")
    plt.show()

    return img, norm
```
